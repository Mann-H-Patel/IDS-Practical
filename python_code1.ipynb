```python
#19


# ============================================================
# PROJECT: NETWORK TRAFFIC CAPTURE AND ANALYSIS USING WIRESHARK
# ============================================================

# -------------------- INSTRUCTIONS ---------------------------
#
# STEP 1: Install Python from:
#         https://www.python.org/downloads/
#
# STEP 2: Install Wireshark from:
#         https://www.wireshark.org/download.html
#
# STEP 3: Windows users should install Npcap:
#         https://npcap.com/
#
# STEP 4: Open Command Prompt and install Scapy:
#         pip install scapy
#
# STEP 5: Save this program as:
#         network_analysis.py
#
# STEP 6: Open Command Prompt in the saved folder.
#
# STEP 7: Run the program using:
#         python network_analysis.py
#
# STEP 8: Wait until 20 packets are captured.
#
# STEP 9: A file named network_capture.pcap will be created.
#
# STEP 10: Open Wireshark.
#
# STEP 11: Select File -> Open.
#
# STEP 12: Open network_capture.pcap.
#
# STEP 13: Use these Wireshark display filters:
#          tcp  - TCP packets
#          udp  - UDP packets
#          icmp - ICMP packets
#          dns  - DNS packets
#          tls  - TLS packets
#
# NOTE: Capture only traffic you are authorized to monitor.
#       On Windows, run with required capture permissions.
#
# -------------------- PYTHON PROGRAM -------------------------

from scapy.all import sniff, IP, TCP, UDP, ICMP, wrpcap
from collections import Counter

# Number of packets to capture
PACKET_COUNT = 20

# Output file name
FILE_NAME = "network_capture.pcap"

# Store protocol counts
protocol_count = Counter()


# Function to analyze captured packets
def analyze_packet(packet):

    if IP in packet:

        source_ip = packet[IP].src
        destination_ip = packet[IP].dst
        packet_length = len(packet)

        # Identify network protocol
        if TCP in packet:
            protocol = "TCP"
            source_port = packet[TCP].sport
            destination_port = packet[TCP].dport

        elif UDP in packet:
            protocol = "UDP"
            source_port = packet[UDP].sport
            destination_port = packet[UDP].dport

        elif ICMP in packet:
            protocol = "ICMP"
            source_port = "-"
            destination_port = "-"

        else:
            protocol = "OTHER"
            source_port = "-"
            destination_port = "-"

        # Update protocol count
        protocol_count[protocol] += 1

        # Display packet details
        print("\n------------------------------")
        print("Source IP:", source_ip)
        print("Destination IP:", destination_ip)
        print("Protocol:", protocol)
        print("Source Port:", source_port)
        print("Destination Port:", destination_port)
        print("Packet Length:", packet_length, "bytes")


# Main function
def main():

    print("====================================")
    print(" NETWORK TRAFFIC CAPTURE AND ANALYSIS")
    print("====================================")

    print("\nStarting packet capture...")
    print("Capturing", PACKET_COUNT, "packets...")

    try:

        # Capture packets from authorized interfaces
        packets = sniff(
            count=PACKET_COUNT,
            filter="ip",
            prn=analyze_packet,
            store=True
        )

        # Save captured packets
        wrpcap(FILE_NAME, packets)

        # Display final analysis
        print("\n====================================")
        print("          ANALYSIS RESULT")
        print("====================================")

        print("Total Packets:", len(packets))

        for protocol, count in protocol_count.items():
            print(protocol, "Packets:", count)

        print("\nCapture completed successfully!")
        print("Saved File:", FILE_NAME)

        print("\nOpen the PCAP file in Wireshark.")

    except PermissionError:

        print("\nERROR: Permission denied.")
        print("Run with the required capture permissions.")

    except Exception as error:

        print("\nERROR:", error)
        print("Check Npcap, Scapy and capture permissions.")


# Start the program
if __name__ == "__main__":
    main()


# -------------------- END OF PROJECT -------------------------
```

```python
#20


# ==========================================================
# PROJECT: ANALYSIS OF SECURE AND INSECURE PROTOCOLS
# USING PYTHON AND WIRESHARK
# ==========================================================

# INSTALLATION INSTRUCTIONS:
#
# 1. Install Python:
#    https://www.python.org/downloads/
#
# 2. Install Wireshark:
#    https://www.wireshark.org/download.html
#
# 3. Install Scapy using Command Prompt:
#    pip install scapy
#
# 4. Run the previous network capture program,
#    or capture traffic using Wireshark.
#
# 5. Save the capture file as:
#    network_capture.pcap
#
# 6. Save this program as:
#    protocol_analysis.py
#
# 7. Run:
#    python protocol_analysis.py
#
# 8. The program displays protocol counts and
#    their security classifications.
#
# 9. Open the PCAP file in Wireshark for
#    detailed packet inspection.
#
# IMPORTANT:
# This program classifies protocol types, not
# the actual security of every connection.
# Capture only authorized network traffic.

from scapy.all import rdpcap, IP, TCP, UDP
from collections import Counter

FILE_NAME = "network_capture.pcap"

# Secure protocol ports
secure_ports = {
    443: "HTTPS / TLS",
    22: "SSH",
    990: "FTPS",
    993: "IMAPS",
    995: "POP3S",
    465: "SMTPS",
    989: "FTPS"
}

# Insecure or unencrypted protocol ports
insecure_ports = {
    80: "HTTP",
    21: "FTP",
    23: "Telnet",
    25: "SMTP",
    110: "POP3",
    143: "IMAP"
}

secure_count = Counter()
insecure_count = Counter()
unknown_count = Counter()

def analyze_protocol(packet):

    if IP not in packet:
        return

    protocol_name = "Unknown"
    category = "Unknown"

    # Check TCP ports
    if TCP in packet:

        src_port = packet[TCP].sport
        dst_port = packet[TCP].dport

        ports = [src_port, dst_port]

        for port in ports:

            if port in secure_ports:
                protocol_name = secure_ports[port]
                category = "Secure"
                break

            elif port in insecure_ports:
                protocol_name = insecure_ports[port]
                category = "Insecure"
                break

    # Check UDP ports
    elif UDP in packet:

        src_port = packet[UDP].sport
        dst_port = packet[UDP].dport

        if 443 in [src_port, dst_port]:
            protocol_name = "QUIC / HTTP3"
            category = "Secure transport (verify)"

        elif 53 in [src_port, dst_port]:
            protocol_name = "DNS"
            category = "Unencrypted by default"

    # Count and display
    if category == "Secure":
        secure_count[protocol_name] += 1

    elif category == "Insecure":
        insecure_count[protocol_name] += 1

    else:
        unknown_count[protocol_name] += 1

    print("-----------------------------------")
    print("Source IP:", packet[IP].src)
    print("Destination IP:", packet[IP].dst)
    print("Protocol:", protocol_name)
    print("Classification:", category)
    print("Packet Length:", len(packet), "bytes")


def main():

    print("===================================")
    print(" SECURE AND INSECURE PROTOCOL ANALYSIS")
    print("===================================")

    try:

        packets = rdpcap(FILE_NAME)

        print("\nTotal Packets:", len(packets))
        print("\nAnalyzing captured traffic...\n")

        for packet in packets:
            analyze_protocol(packet)

        print("\n===================================")
        print("          FINAL REPORT")
        print("===================================")

        print("\nSECURE PROTOCOLS")

        for name, count in secure_count.items():
            print(name, ":", count, "packets")

        print("\nINSECURE PROTOCOLS")

        for name, count in insecure_count.items():
            print(name, ":", count, "packets")

        print("\nOTHER / UNCLASSIFIED PROTOCOLS")

        for name, count in unknown_count.items():
            print(name, ":", count, "packets")

        print("\nAnalysis completed successfully!")

    except FileNotFoundError:

        print("\nERROR: PCAP file not found.")
        print("Place network_capture.pcap in this folder.")

    except Exception as error:

        print("\nERROR:", error)


if __name__ == "__main__":
    main()

# ==========================================================
# END OF PROGRAM
# ==========================================================
```

```python
#21


# ============================================================
# PROJECT: ROLE-BASED ACCESS CONTROL (RBAC)
# USING PYTHON AND SQLITE DATABASE
# ============================================================

# -------------------- INSTRUCTIONS ---------------------------
#
# STEP 1: Install Python 3.
#        https://www.python.org/downloads/
#
# STEP 2: Open VS Code, PyCharm or Notepad.
#
# STEP 3: Copy this complete program.
#
# STEP 4: Save the file as:
#        rbac_system.py
#
# STEP 5: Open Command Prompt in the saved folder.
#
# STEP 6: Run the program using:
#        python rbac_system.py
#
# STEP 7: Select a sample username to log in.
#
# STEP 8: Select an operation from the menu.
#
# STEP 9: The system checks the user's role and
#        allows or denies access automatically.
#
# NOTE:
# SQLite is included with Python.
# No additional packages are required.
#
# SAMPLE LOGIN ACCOUNTS:
# Admin    : admin
# Manager  : manager
# Employee : employee
#
# ============================================================

import sqlite3

# Connect to SQLite database
conn = sqlite3.connect("rbac_database.db")
cursor = conn.cursor()

# Create records table
cursor.execute("""
CREATE TABLE IF NOT EXISTS records (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    department TEXT NOT NULL
)
""")

# Insert sample records only if table is empty
cursor.execute("SELECT COUNT(*) FROM records")

if cursor.fetchone()[0] == 0:
    cursor.executemany(
        "INSERT INTO records (name, department) VALUES (?, ?)",
        [
            ("Rahul", "IT"),
            ("Priya", "HR"),
            ("Amit", "Finance")
        ]
    )

conn.commit()

# Define roles and permissions
roles = {
    "Admin": ["view", "add", "update", "delete"],
    "Manager": ["view", "update"],
    "Employee": ["view"]
}

# Sample users and assigned roles
users = {
    "admin": "Admin",
    "manager": "Manager",
    "employee": "Employee"
}

# Function to check permissions
def check_permission(role, action):

    if action in roles.get(role, []):
        return True

    return False


# Function to view records
def view_records():

    cursor.execute("SELECT * FROM records")
    records = cursor.fetchall()

    print("\n========== DATABASE RECORDS ==========")

    if not records:
        print("No records available.")

    for record in records:
        print(record)

    print("=======================================")


# Function to add records
def add_record():

    name = input("Enter employee name: ").strip()
    department = input("Enter department: ").strip()

    if not name or not department:
        print("Name and department cannot be empty.")
        return

    cursor.execute(
        "INSERT INTO records (name, department) VALUES (?, ?)",
        (name, department)
    )

    conn.commit()
    print("Record added successfully!")


# Function to update records
def update_record():

    try:
        record_id = int(input("Enter record ID: "))
    except ValueError:
        print("Please enter a valid ID.")
        return

    name = input("Enter new name: ").strip()
    department = input("Enter new department: ").strip()

    if not name or not department:
        print("Fields cannot be empty.")
        return

    cursor.execute(
        "UPDATE records SET name=?, department=? WHERE id=?",
        (name, department, record_id)
    )

    conn.commit()

    if cursor.rowcount > 0:
        print("Record updated successfully!")
    else:
        print("Record not found.")


# Function to delete records
def delete_record():

    try:
        record_id = int(input("Enter record ID to delete: "))
    except ValueError:
        print("Please enter a valid ID.")
        return

    cursor.execute(
        "DELETE FROM records WHERE id=?",
        (record_id,)
    )

    conn.commit()

    if cursor.rowcount > 0:
        print("Record deleted successfully!")
    else:
        print("Record not found.")


# Function to display menu
def display_menu():

    print("\n========== RBAC MENU ==========")
    print("1. View Records")
    print("2. Add Record")
    print("3. Update Record")
    print("4. Delete Record")
    print("5. Logout")
    print("===============================")


# Main application
def main():

    print("\n================================")
    print(" ROLE-BASED ACCESS CONTROL SYSTEM")
    print("================================")

    print("\nAvailable Users:")
    print("1. admin")
    print("2. manager")
    print("3. employee")

    username = input("\nEnter username: ").strip().lower()

    # Authenticate sample username
    if username not in users:
        print("Access denied! Invalid username.")
        return

    role = users[username]

    print("\nLogin successful!")
    print("Username:", username)
    print("Assigned Role:", role)

    while True:

        display_menu()

        choice = input("Enter your choice: ").strip()

        if choice == "1":
            action = "view"

        elif choice == "2":
            action = "add"

        elif choice == "3":
            action = "update"

        elif choice == "4":
            action = "delete"

        elif choice == "5":
            print("Logged out successfully!")
            break

        else:
            print("Invalid choice. Try again.")
            continue

        # Check user permission before execution
        if check_permission(role, action):

            print("\nAccess Granted!")

            if action == "view":
                view_records()

            elif action == "add":
                add_record()

            elif action == "update":
                update_record()

            elif action == "delete":
                delete_record()

        else:
            print("\nACCESS DENIED!")
            print("Your role does not have permission for:", action)

    print("\nThank you for using the RBAC system.")


# Execute program
if __name__ == "__main__":

    try:
        main()

    finally:
        conn.close()

# ============================================================
# END OF RBAC PROJECT
# ============================================================
```

```python
#22


# ============================================================
# PROJECT: ORGANIZATIONAL DATA ACCESS CONTROL SYSTEM
# USING PYTHON AND SQLITE
# ============================================================

# -------------------- INSTRUCTIONS ---------------------------
#
# STEP 1: Install Python 3.
#         https://www.python.org/downloads/
#
# STEP 2: Open VS Code, PyCharm or Notepad.
#
# STEP 3: Copy this complete program.
#
# STEP 4: Save it as:
#         organizational_access.py
#
# STEP 5: Open Command Prompt in the saved folder.
#
# STEP 6: Run the program:
#         python organizational_access.py
#
# STEP 7: Select a sample username.
#
# STEP 8: Choose a data resource and operation.
#
# STEP 9: The system checks permissions and
#         grants or denies access.
#
# STEP 10: The database is automatically created.
#
# SAMPLE USERS:
# admin     - Administrator
# manager   - Manager
# hr        - HR Employee
# employee  - General Employee
#
# NOTE: This is an educational demonstration.
# Real systems require secure authentication.
#
# ============================================================

import sqlite3

# Connect to organizational database
conn = sqlite3.connect("organization.db")
cursor = conn.cursor()

# Create organizational resources table
cursor.execute("""
CREATE TABLE IF NOT EXISTS resources (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    resource_name TEXT NOT NULL,
    category TEXT NOT NULL,
    content TEXT NOT NULL
)
""")

# Insert sample data if database is empty
cursor.execute("SELECT COUNT(*) FROM resources")

if cursor.fetchone()[0] == 0:

    sample_data = [
        ("Company Notice", "Public", "Annual meeting announcement"),
        ("Employee Records", "Employee", "Employee department information"),
        ("Salary Report", "Financial", "Monthly salary report"),
        ("Company Policy", "Public", "Organizational rules and policies"),
        ("Staff Details", "Employee", "Staff contact and role information")
    ]

    cursor.executemany("""
    INSERT INTO resources
    (resource_name, category, content)
    VALUES (?, ?, ?)
    """, sample_data)

    conn.commit()


# Define organizational roles and permissions
permissions = {

    "Admin": {
        "Public": ["read", "write", "delete"],
        "Employee": ["read", "write", "delete"],
        "Financial": ["read", "write", "delete"]
    },

    "Manager": {
        "Public": ["read"],
        "Employee": ["read", "write"],
        "Financial": ["read"]
    },

    "HR": {
        "Public": ["read"],
        "Employee": ["read", "write"],
        "Financial": []
    },

    "Employee": {
        "Public": ["read"],
        "Employee": [],
        "Financial": []
    }
}


# Sample user-role assignments
users = {
    "admin": "Admin",
    "manager": "Manager",
    "hr": "HR",
    "employee": "Employee"
}


# Check access permissions
def check_access(role, category, action):

    allowed_actions = permissions.get(role, {}).get(category, [])

    return action in allowed_actions


# Display available resources
def view_resources():

    cursor.execute("""
    SELECT id, resource_name, category
    FROM resources
    """)

    records = cursor.fetchall()

    print("\n========== AVAILABLE RESOURCES ==========")

    for record in records:
        print(record)

    print("==========================================")


# Read resource
def read_resource(category):

    cursor.execute("""
    SELECT id, resource_name, content
    FROM resources
    WHERE category = ?
    """, (category,))

    records = cursor.fetchall()

    print("\nRESOURCE INFORMATION")

    for record in records:
        print("\nID:", record[0])
        print("Name:", record[1])
        print("Content:", record[2])


# Add resource
def add_resource(category):

    name = input("Enter resource name: ").strip()
    content = input("Enter resource content: ").strip()

    if not name or not content:
        print("Fields cannot be empty.")
        return

    cursor.execute("""
    INSERT INTO resources
    (resource_name, category, content)
    VALUES (?, ?, ?)
    """, (name, category, content))

    conn.commit()

    print("Resource added successfully!")


# Delete resource
def delete_resource(category):

    try:
        resource_id = int(input("Enter resource ID: "))
    except ValueError:
        print("Invalid resource ID.")
        return

    cursor.execute("""
    DELETE FROM resources
    WHERE id = ? AND category = ?
    """, (resource_id, category))

    conn.commit()

    if cursor.rowcount > 0:
        print("Resource deleted successfully!")
    else:
        print("Resource not found.")


# Main application
def main():

    print("\n========================================")
    print(" ORGANIZATIONAL DATA ACCESS CONTROL")
    print("========================================")

    print("\nAvailable Users:")
    print("1. admin")
    print("2. manager")
    print("3. hr")
    print("4. employee")

    username = input("\nEnter username: ").strip().lower()

    if username not in users:
        print("Access Denied: Invalid user.")
        return

    role = users[username]

    print("\nLogin Successful!")
    print("User:", username)
    print("Role:", role)

    while True:

        print("\n========== MAIN MENU ==========")
        print("1. View Available Resources")
        print("2. Read Public Data")
        print("3. Read Employee Data")
        print("4. Read Financial Data")
        print("5. Add Public Resource")
        print("6. Add Employee Resource")
        print("7. Add Financial Resource")
        print("8. Delete Public Resource")
        print("9. Delete Employee Resource")
        print("10. Delete Financial Resource")
        print("11. Logout")
        print("===============================")

        choice = input("Enter your choice: ").strip()

        operations = {
            "2": ("Public", "read"),
            "3": ("Employee", "read"),
            "4": ("Financial", "read"),
            "5": ("Public", "write"),
            "6": ("Employee", "write"),
            "7": ("Financial", "write"),
            "8": ("Public", "delete"),
            "9": ("Employee", "delete"),
            "10": ("Financial", "delete")
        }

        if choice == "1":
            view_resources()
            continue

        if choice == "11":
            print("Logged out successfully!")
            break

        if choice not in operations:
            print("Invalid choice.")
            continue

        category, action = operations[choice]

        # Verify permission before accessing data
        if check_access(role, category, action):

            print("\nACCESS GRANTED!")

            if action == "read":
                read_resource(category)

            elif action == "write":
                add_resource(category)

            elif action == "delete":
                delete_resource(category)

        else:

            print("\nACCESS DENIED!")
            print("Role:", role)
            print("Resource:", category)
            print("Action:", action)
            print("You do not have permission.")


# Execute application
if __name__ == "__main__":

    try:
        main()

    finally:
        conn.close()

# ============================================================
# END OF PROJECT
# ============================================================
```

```python
#23


# ============================================================
# PROJECT: SECURE ACCESS PERMISSIONS IN CLOUD STORAGE
# USING PYTHON
# ============================================================

# -------------------- INSTRUCTIONS ---------------------------
#
# STEP 1: Install Python 3.
#         https://www.python.org/downloads/
#
# STEP 2: Open VS Code, PyCharm or Notepad.
#
# STEP 3: Copy this complete Python program.
#
# STEP 4: Save the file as:
#         cloud_storage_security.py
#
# STEP 5: Open Command Prompt in the saved folder.
#
# STEP 6: Run the program:
#         python cloud_storage_security.py
#
# STEP 7: Select a username:
#         admin
#         manager
#         employee
#
# STEP 8: Select an operation from the menu.
#
# STEP 9: The system checks the user's permissions.
#
# STEP 10: Files are stored in a folder named:
#          cloud_storage
#
# STEP 11: Try different users to test access control.
#
# REQUIREMENTS:
# - Python 3
# - No additional libraries required
#
# NOTE:
# This is a local cloud-storage simulation.
# It does not connect to AWS, Azure or Google Cloud.
# Real cloud storage requires secure authentication,
# encryption and provider-side access policies.
#
# ============================================================

import os
from pathlib import Path
from datetime import datetime

# Cloud storage directory
STORAGE = Path("cloud_storage")
STORAGE.mkdir(exist_ok=True)

# Create sample files if they do not exist
sample_files = {
    "company_notice.txt": "Welcome to the company cloud storage.",
    "project_report.txt": "Project progress report.",
    "employee_data.txt": "Sample employee information."
}

for filename, content in sample_files.items():
    file_path = STORAGE / filename

    if not file_path.exists():
        file_path.write_text(content, encoding="utf-8")


# Sample users and their roles
users = {
    "admin": "Admin",
    "manager": "Manager",
    "employee": "Employee"
}


# Define role permissions
permissions = {

    "Admin": ["view", "upload", "delete"],

    "Manager": ["view", "upload"],

    "Employee": ["view"]
}


# Audit log
def log_activity(username, action, result):

    with open("access_log.txt", "a", encoding="utf-8") as log:

        log.write(
            f"{datetime.now()} | "
            f"User: {username} | "
            f"Action: {action} | "
            f"Result: {result}\n"
        )


# Check permission
def check_permission(role, action):

    return action in permissions.get(role, [])


# Validate file name
def get_safe_path(filename):

    filename = filename.strip()

    if not filename or filename in (".", ".."):
        return None

    # Reject directory traversal and path separators
    if Path(filename).name != filename:
        return None

    if "/" in filename or "\\" in filename:
        return None

    return STORAGE / filename


# View stored files
def view_files():

    print("\n========== CLOUD STORAGE FILES ==========")

    files = [
        file for file in STORAGE.iterdir()
        if file.is_file()
    ]

    if not files:
        print("No files available.")

    for file in files:
        print(file.name)

    print("=========================================")


# Upload file
def upload_file():

    filename = input("Enter file name: ").strip()
    file_path = get_safe_path(filename)

    if file_path is None:
        print("Invalid file name.")
        return

    if file_path.exists():
        print("File already exists. Choose another name.")
        return

    content = input("Enter file content: ")

    try:
        with file_path.open("x", encoding="utf-8") as file:
            file.write(content)

        print("File uploaded successfully!")

    except OSError as error:
        print("Upload failed:", error)


# Delete file
def delete_file():

    filename = input("Enter file name to delete: ").strip()
    file_path = get_safe_path(filename)

    if file_path is None:
        print("Invalid file name.")
        return

    if file_path.is_file():

        file_path.unlink()
        print("File deleted successfully!")

    else:
        print("File not found.")


# Read file
def read_file():

    filename = input("Enter file name to view: ").strip()
    file_path = get_safe_path(filename)

    if file_path is None:
        print("Invalid file name.")
        return

    if file_path.is_file():

        print("\n========== FILE CONTENT ==========")

        print(file_path.read_text(encoding="utf-8"))

        print("==================================")

    else:
        print("File not found.")


# Main application
def main():

    print("\n======================================")
    print(" SECURE CLOUD STORAGE ACCESS SYSTEM")
    print("======================================")

    print("\nAvailable Users:")
    print("1. admin")
    print("2. manager")
    print("3. employee")

    username = input("\nEnter username: ").strip().lower()

    # User validation
    if username not in users:

        print("Access Denied! Invalid username.")
        return

    role = users[username]

    print("\nLogin Successful!")
    print("Username:", username)
    print("Assigned Role:", role)

    while True:

        print("\n========== CLOUD STORAGE MENU ==========")
        print("1. View Files")
        print("2. Read File")
        print("3. Upload File")
        print("4. Delete File")
        print("5. Logout")
        print("========================================")

        choice = input("Enter your choice: ").strip()

        actions = {
            "1": "view",
            "2": "view",
            "3": "upload",
            "4": "delete"
        }

        if choice == "5":

            print("Logged out successfully!")
            log_activity(username, "Logout", "Success")
            break

        if choice not in actions:

            print("Invalid choice.")
            continue

        action = actions[choice]

        # Check access permission
        if check_permission(role, action):

            print("\nACCESS GRANTED!")

            try:

                if choice == "1":
                    view_files()

                elif choice == "2":
                    read_file()

                elif choice == "3":
                    upload_file()

                elif choice == "4":
                    delete_file()

                log_activity(username, action, "Allowed")

            except OSError as error:

                print("Operation failed:", error)
                log_activity(username, action, "Failed")

        else:

            print("\nACCESS DENIED!")
            print("User Role:", role)
            print("Requested Action:", action)
            print("You do not have permission.")

            log_activity(username, action, "Denied")


# Run application
if __name__ == "__main__":

    main()

# ============================================================
# END OF CLOUD STORAGE SECURITY PROJECT
# ============================================================
```

```python
#25


# ============================================================
# PRACTICAL: ANALYZE DATABASE SECURITY MECHANISMS AND PRIVILEGES
# ============================================================
#
# AIM:
# Demonstrate database security using user roles, permissions,
# access control, parameterized SQL queries and audit logging.
#
# REQUIREMENTS:
# 1. Python 3
# 2. No external libraries required.
#
# HOW TO RUN:
# 1. Save this file as database_security.py
# 2. Open a terminal in the folder containing this file.
# 3. Run: python database_security.py
# 4. Select a username to log in:
#       admin   - all permissions
#       manager - view, insert and update
#       viewer  - view only
# 5. Select operations from the menu.
#
# NOTE:
# This is an educational simulation. Login uses usernames only.
# It is NOT suitable for production use.
# ============================================================

import sqlite3
from datetime import datetime

DATABASE = "database_security.db"

# ------------------------------------------------------------
# CONNECT TO DATABASE
# ------------------------------------------------------------

connection = sqlite3.connect(DATABASE)
connection.execute("PRAGMA foreign_keys = ON")
cursor = connection.cursor()

# ------------------------------------------------------------
# CREATE TABLES
# ------------------------------------------------------------

cursor.execute("""
CREATE TABLE IF NOT EXISTS users (
    username TEXT PRIMARY KEY,
    role TEXT NOT NULL
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS privileges (
    role TEXT NOT NULL,
    operation TEXT NOT NULL,
    PRIMARY KEY (role, operation)
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS records (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    department TEXT NOT NULL
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS audit_log (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    username TEXT NOT NULL,
    operation TEXT NOT NULL,
    status TEXT NOT NULL,
    event_time TEXT NOT NULL
)
""")

# ------------------------------------------------------------
# CREATE SAMPLE USERS AND ROLES
# ------------------------------------------------------------

users = [
    ("admin", "Admin"),
    ("manager", "Manager"),
    ("viewer", "Viewer")
]

cursor.executemany("""
INSERT OR IGNORE INTO users (username, role)
VALUES (?, ?)
""", users)

# ------------------------------------------------------------
# ASSIGN PRIVILEGES
# ------------------------------------------------------------

role_privileges = [
    ("Admin", "VIEW"),
    ("Admin", "INSERT"),
    ("Admin", "UPDATE"),
    ("Admin", "DELETE"),

    ("Manager", "VIEW"),
    ("Manager", "INSERT"),
    ("Manager", "UPDATE"),

    ("Viewer", "VIEW")
]

cursor.executemany("""
INSERT OR IGNORE INTO privileges (role, operation)
VALUES (?, ?)
""", role_privileges)

# ------------------------------------------------------------
# INSERT SAMPLE RECORDS
# ------------------------------------------------------------

cursor.execute("SELECT COUNT(*) FROM records")

if cursor.fetchone()[0] == 0:
    sample_records = [
        ("Aarav", "IT"),
        ("Diya", "HR"),
        ("Rahul", "Finance")
    ]

    cursor.executemany("""
    INSERT INTO records (name, department)
    VALUES (?, ?)
    """, sample_records)

connection.commit()

# ------------------------------------------------------------
# AUDIT LOGGING
# ------------------------------------------------------------

def log_activity(username, operation, status):
    event_time = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    cursor.execute("""
    INSERT INTO audit_log
    (username, operation, status, event_time)
    VALUES (?, ?, ?, ?)
    """, (username, operation, status, event_time))

    connection.commit()

# ------------------------------------------------------------
# CHECK USER PRIVILEGES
# ------------------------------------------------------------

def check_permission(username, operation):

    cursor.execute("""
    SELECT role FROM users
    WHERE username = ?
    """, (username,))

    result = cursor.fetchone()

    if result is None:
        return False

    role = result[0]

    cursor.execute("""
    SELECT 1 FROM privileges
    WHERE role = ? AND operation = ?
    """, (role, operation))

    return cursor.fetchone() is not None

# ------------------------------------------------------------
# VIEW RECORDS
# ------------------------------------------------------------

def view_records(username):

    if not check_permission(username, "VIEW"):
        print("Access Denied: You cannot view records.")
        log_activity(username, "VIEW", "DENIED")
        return

    cursor.execute("SELECT * FROM records")
    records = cursor.fetchall()

    print("\n--- DATABASE RECORDS ---")

    for record in records:
        print(
            "ID:", record[0],
            "| Name:", record[1],
            "| Department:", record[2]
        )

    log_activity(username, "VIEW", "ALLOWED")

# ------------------------------------------------------------
# INSERT RECORD
# ------------------------------------------------------------

def insert_record(username):

    if not check_permission(username, "INSERT"):
        print("Access Denied: You cannot insert records.")
        log_activity(username, "INSERT", "DENIED")
        return

    name = input("Enter name: ").strip()
    department = input("Enter department: ").strip()

    if not name or not department:
        print("Name and department cannot be empty.")
        log_activity(username, "INSERT", "FAILED")
        return

    # Parameterized query helps prevent SQL injection.
    cursor.execute("""
    INSERT INTO records (name, department)
    VALUES (?, ?)
    """, (name, department))

    connection.commit()

    print("Record inserted successfully.")
    log_activity(username, "INSERT", "ALLOWED")

# ------------------------------------------------------------
# UPDATE RECORD
# ------------------------------------------------------------

def update_record(username):

    if not check_permission(username, "UPDATE"):
        print("Access Denied: You cannot update records.")
        log_activity(username, "UPDATE", "DENIED")
        return

    try:
        record_id = int(input("Enter record ID to update: "))
    except ValueError:
        print("Invalid record ID.")
        log_activity(username, "UPDATE", "FAILED")
        return

    name = input("Enter new name: ").strip()
    department = input("Enter new department: ").strip()

    if not name or not department:
        print("Name and department cannot be empty.")
        log_activity(username, "UPDATE", "FAILED")
        return

    cursor.execute("""
    UPDATE records
    SET name = ?, department = ?
    WHERE id = ?
    """, (name, department, record_id))

    connection.commit()

    if cursor.rowcount == 0:
        print("Record not found.")
        log_activity(username, "UPDATE", "FAILED")
    else:
        print("Record updated successfully.")
        log_activity(username, "UPDATE", "ALLOWED")

# ------------------------------------------------------------
# DELETE RECORD
# ------------------------------------------------------------

def delete_record(username):

    if not check_permission(username, "DELETE"):
        print("Access Denied: You cannot delete records.")
        log_activity(username, "DELETE", "DENIED")
        return

    try:
        record_id = int(input("Enter record ID to delete: "))
    except ValueError:
        print("Invalid record ID.")
        log_activity(username, "DELETE", "FAILED")
        return

    cursor.execute("""
    DELETE FROM records WHERE id = ?
    """, (record_id,))

    connection.commit()

    if cursor.rowcount == 0:
        print("Record not found.")
        log_activity(username, "DELETE", "FAILED")
    else:
        print("Record deleted successfully.")
        log_activity(username, "DELETE", "ALLOWED")

# ------------------------------------------------------------
# DISPLAY USER PRIVILEGES
# ------------------------------------------------------------

def show_privileges(username):

    cursor.execute("""
    SELECT role FROM users WHERE username = ?
    """, (username,))

    result = cursor.fetchone()

    if result is None:
        return

    role = result[0]

    cursor.execute("""
    SELECT operation FROM privileges
    WHERE role = ?
    ORDER BY operation
    """, (role,))

    permissions = [row[0] for row in cursor.fetchall()]

    print("\nRole:", role)
    print("Assigned privileges:", ", ".join(permissions))

# ------------------------------------------------------------
# DISPLAY AUDIT LOG
# ------------------------------------------------------------

def show_audit_log(username):

    if not check_permission(username, "VIEW"):
        print("Access Denied.")
        log_activity(username, "AUDIT_LOG", "DENIED")
        return

    cursor.execute("""
    SELECT username, operation, status, event_time
    FROM audit_log
    ORDER BY id DESC
    LIMIT 20
    """)

    print("\n--- RECENT AUDIT LOG ---")

    rows = cursor.fetchall()

    if not rows:
        print("No activity recorded yet.")

    for row in rows:
        print(
            "User:", row[0],
            "| Operation:", row[1],
            "| Status:", row[2],
            "| Time:", row[3]
        )

    log_activity(username, "AUDIT_LOG", "ALLOWED")

# ------------------------------------------------------------
# LOGIN
# ------------------------------------------------------------

print("======================================")
print("       DATABASE SECURITY SYSTEM")
print("======================================")

print("\nAvailable usernames:")
print("1. admin")
print("2. manager")
print("3. viewer")

username = input("\nEnter username: ").strip()

cursor.execute("""
SELECT username FROM users WHERE username = ?
""", (username,))

if cursor.fetchone() is None:
    print("Invalid username. Access denied.")
    connection.close()
    raise SystemExit

print("\nLogin successful.")

show_privileges(username)

# ------------------------------------------------------------
# MAIN MENU
# ------------------------------------------------------------

while True:

    print("\n========== MAIN MENU ==========")
    print("1. View Records")
    print("2. Insert Record")
    print("3. Update Record")
    print("4. Delete Record")
    print("5. View My Privileges")
    print("6. View Audit Log")
    print("7. Exit")

    choice = input("Enter your choice: ").strip()

    if choice == "1":
        view_records(username)

    elif choice == "2":
        insert_record(username)

    elif choice == "3":
        update_record(username)

    elif choice == "4":
        delete_record(username)

    elif choice == "5":
        show_privileges(username)

    elif choice == "6":
        show_audit_log(username)

    elif choice == "7":
        print("Logged out successfully.")
        break

    else:
        print("Invalid choice. Try again.")

connection.close()

print("\nDatabase connection closed.")
print("Program completed successfully.")
```

```python
#26


# ============================================================
# PRACTICAL: MONITOR AND ANALYZE SYSTEM LOGS
# TO IDENTIFY SECURITY EVENTS
# ============================================================
#
# AIM:
# Analyze system logs and identify possible security events.
#
# REQUIREMENTS:
# 1. Python 3
# 2. No external libraries required.
#
# HOW TO RUN:
# 1. Save this file as system_log_analyzer.py
# 2. Open a terminal in the same folder.
# 3. Run: python system_log_analyzer.py
# 4. The program creates a sample_system.log file.
# 5. It analyzes the log and generates security_report.txt.
#
# NOTE:
# This practical uses fictional sample logs.
# It does not monitor anyone else's computer or network.
# ============================================================

import csv
import os
from collections import Counter, defaultdict
from datetime import datetime

LOG_FILE = "sample_system.log"
REPORT_FILE = "security_report.txt"

# ------------------------------------------------------------
# STEP 1: CREATE SAMPLE SYSTEM LOGS
# ------------------------------------------------------------

def create_sample_logs():

    if os.path.exists(LOG_FILE):
        return

    sample_logs = [
        ["2026-10-31 09:00:00", "admin", "LOGIN_SUCCESS",
         "192.168.1.10", "Admin logged in"],

        ["2026-10-31 09:05:00", "user1", "LOGIN_FAILED",
         "192.168.1.20", "Incorrect password"],

        ["2026-10-31 09:06:00", "user1", "LOGIN_FAILED",
         "192.168.1.20", "Incorrect password"],

        ["2026-10-31 09:07:00", "user1", "LOGIN_FAILED",
         "192.168.1.20", "Incorrect password"],

        ["2026-10-31 09:10:00", "manager", "LOGIN_SUCCESS",
         "192.168.1.30", "Manager logged in"],

        ["2026-10-31 09:15:00", "user2", "ACCESS_DENIED",
         "192.168.1.40", "Unauthorized file access"],

        ["2026-10-31 09:20:00", "admin", "FILE_MODIFIED",
         "192.168.1.10", "Configuration file modified"],

        ["2026-10-31 09:25:00", "user3", "LOGIN_FAILED",
         "192.168.1.50", "Incorrect password"],

        ["2026-10-31 09:30:00", "user2", "ACCESS_DENIED",
         "192.168.1.40", "Restricted resource access"]
    ]

    with open(LOG_FILE, "w", newline="") as file:
        writer = csv.writer(file)

        writer.writerow([
            "timestamp", "username", "event",
            "ip_address", "details"
        ])

        writer.writerows(sample_logs)

    print("Sample log file created:", LOG_FILE)


# ------------------------------------------------------------
# STEP 2: READ SYSTEM LOGS
# ------------------------------------------------------------

def read_logs():

    logs = []

    try:
        with open(LOG_FILE, "r", newline="") as file:
            reader = csv.DictReader(file)

            for row in reader:
                try:
                    datetime.strptime(
                        row["timestamp"], "%Y-%m-%d %H:%M:%S"
                    )

                    logs.append(row)

                except (ValueError, KeyError):
                    print("Skipping invalid log entry:", row)

    except FileNotFoundError:
        print("Log file not found.")

    return logs


# ------------------------------------------------------------
# STEP 3: ANALYZE SECURITY EVENTS
# ------------------------------------------------------------

def analyze_logs(logs):

    failed_logins = Counter()
    denied_access = Counter()

    successful_logins = 0
    file_modifications = 0

    suspicious_events = []
    security_events = []

    for log in logs:

        event = log["event"]
        username = log["username"]
        ip = log["ip_address"]
        timestamp = log["timestamp"]
        details = log["details"]

        if event == "LOGIN_FAILED":

            failed_logins[ip] += 1

            security_events.append(
                f"[WARNING] Failed login by {username} "
                f"from {ip} at {timestamp}"
            )

        elif event == "LOGIN_SUCCESS":

            successful_logins += 1

            security_events.append(
                f"[INFO] Successful login by {username} "
                f"from {ip} at {timestamp}"
            )

        elif event == "ACCESS_DENIED":

            denied_access[ip] += 1

            security_events.append(
                f"[WARNING] Access denied for {username} "
                f"from {ip} at {timestamp}"
            )

        elif event == "FILE_MODIFIED":

            file_modifications += 1

            security_events.append(
                f"[INFO] File modification by {username} "
                f"at {timestamp}: {details}"
            )

    # Detect repeated failed login attempts.
    for ip, count in failed_logins.items():

        if count >= 3:

            message = (
                f"[HIGH] {count} failed login attempts "
                f"detected from IP address {ip}"
            )

            suspicious_events.append(message)

    # Detect repeated denied access events.
    for ip, count in denied_access.items():

        if count >= 2:

            message = (
                f"[WARNING] {count} access-denied events "
                f"detected from IP address {ip}"
            )

            suspicious_events.append(message)

    return {
        "total": len(logs),
        "failed": sum(failed_logins.values()),
        "successful": successful_logins,
        "denied": sum(denied_access.values()),
        "modifications": file_modifications,
        "failed_by_ip": failed_logins,
        "denied_by_ip": denied_access,
        "events": security_events,
        "suspicious": suspicious_events
    }


# ------------------------------------------------------------
# STEP 4: DISPLAY SECURITY REPORT
# ------------------------------------------------------------

def display_report(result):

    print("\n========================================")
    print("          SYSTEM SECURITY REPORT")
    print("========================================")

    print("Total log entries:", result["total"])
    print("Successful logins:", result["successful"])
    print("Failed logins:", result["failed"])
    print("Access denied events:", result["denied"])
    print("File modifications:", result["modifications"])

    print("\n--- SECURITY EVENTS ---")

    for event in result["events"]:
        print(event)

    print("\n--- SUSPICIOUS ACTIVITIES ---")

    if result["suspicious"]:

        for event in result["suspicious"]:
            print(event)

    else:
        print("No suspicious activity detected.")


# ------------------------------------------------------------
# STEP 5: SAVE REPORT TO A TEXT FILE
# ------------------------------------------------------------

def save_report(result):

    with open(REPORT_FILE, "w") as file:

        file.write("SYSTEM SECURITY ANALYSIS REPORT\n")
        file.write("=" * 40 + "\n\n")

        file.write(
            f"Total log entries: {result['total']}\n"
        )

        file.write(
            f"Successful logins: {result['successful']}\n"
        )

        file.write(
            f"Failed logins: {result['failed']}\n"
        )

        file.write(
            f"Access denied events: {result['denied']}\n"
        )

        file.write(
            f"File modifications: {result['modifications']}\n"
        )

        file.write("\nSECURITY EVENTS\n")
        file.write("-" * 40 + "\n")

        for event in result["events"]:
            file.write(event + "\n")

        file.write("\nSUSPICIOUS ACTIVITIES\n")
        file.write("-" * 40 + "\n")

        if result["suspicious"]:

            for event in result["suspicious"]:
                file.write(event + "\n")

        else:
            file.write("No suspicious activity detected.\n")

    print("\nReport saved successfully:", REPORT_FILE)


# ------------------------------------------------------------
# MAIN PROGRAM
# ------------------------------------------------------------

def main():

    print("SYSTEM LOG MONITORING AND ANALYSIS")
    print("----------------------------------")

    create_sample_logs()

    logs = read_logs()

    if not logs:
        print("No valid log entries available.")
        return

    result = analyze_logs(logs)

    display_report(result)

    save_report(result)

    print("\nAnalysis completed successfully.")


if __name__ == "__main__":
    main()
```
